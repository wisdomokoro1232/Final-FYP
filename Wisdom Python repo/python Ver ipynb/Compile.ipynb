```python
import numpy as np

def Compile(filenames):
    n= len(filenames)
    for i in range(n):
        if i == 0:
            AllfGyroX, AllfGyroY, AllfGyroZ,  = segment_Gyro(filenames[i])
            AllfAccelX, AllfAccelY, AllfAccelZ = segment_Accel(filenames[i])
        else:
            DfGyroX, DfGyroY, DfGyroZ= segment_Gyro(filenames[i])
            DfAccelX, DfAccelY, DfAccelZ = segment_Accel(filenames[i])
            AllfGyroX = np.hstack((AllfGyroX, DfGyroX))
            AllfGyroY = np.hstack((AllfGyroY, DfGyroY))
            AllfGyroZ = np.hstack((AllfGyroZ, DfGyroZ))
            AllfAccelX = np.hstack((AllfAccelX, DfAccelX))
            AllfAccelY = np.hstack((AllfAccelY, DfAccelY))
            AllfAccelZ = np.hstack((AllfAccelZ, DfAccelZ))

    return AllfGyroX, AllfGyroY, AllfGyroZ, AllfAccelX, AllfAccelY, AllfAccelZ
```

```python
import matplotlib.pyplot as plt

# Assuming AllfGyroX, AllfGyroY, AllfGyroZ, AllfAccelX, AllfAccelY, AllfAccelZ are defined
for i in range(AllfGyroX.shape[1]):
    plt.figure()
    plt.plot(AllfGyroX[:, i], 'r')
    plt.plot(AllfGyroY[:, i], 'g')
    plt.plot(AllfGyroZ[:, i], 'b')
    plt.xlabel('Time (s)')
    plt.ylabel('Gyroscope (°/s)')
    plt.xticks([126, 252, 378, 504], ['1', '2', '3', '4'])

    plt.figure()
    plt.plot(AllfAccelX[:, i], 'r')
    plt.plot(AllfAccelY[:, i], 'g')
    plt.plot(AllfAccelZ[:, i], 'b')
    plt.xlabel('Time (s)')
    plt.ylabel('Acceleration (g)')
    plt.xticks([126, 252, 378, 504], ['1', '2', '3', '4'])

plt.show()
```
