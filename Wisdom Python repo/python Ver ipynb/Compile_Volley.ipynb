```python
import numpy as np

def Compile_Volley(filenames):
    n= len(filenames)
    for i in range(n):
        if i == 0:
            AllfGyroX, AllfGyroY, AllfGyroZ,  = segment_Gyro(filenames[i])
            AllfAccelX, AllfAccelY, AllfAccelZ = segment_Accel(filenames[i])
        else:
            DfGyroX, DfGyroY, DfGyroZ= segment_Gyro(filenames[i])
            DfAccelX, DfAccelY, DfAccelZ = segment_Accel(filenames[i])
            AllfGyroX = np.hstack((AllfGyroX, DfGyroX))
            AllfGyroY = np.hstack((AllfGyroY, DfGyroY))
            AllfGyroZ = np.hstack((AllfGyroZ, DfGyroZ))
            AllfAccelX = np.hstack((AllfAccelX, DfAccelX))
            AllfAccelY = np.hstack((AllfAccelY, DfAccelY))
            AllfAccelZ = np.hstack((AllfAccelZ, DfAccelZ))

    return AllfGyroX, AllfGyroY, AllfGyroZ, AllfAccelX, AllfAccelY, AllfAccelZ
```

```python
import matplotlib.pyplot as plt

def plot_sensor_data(AllfGyroX, AllfGyroY, AllfGyroZ, AllfAccelX, AllfAccelY, AllfAccelZ):
    """
    Plots the gyroscope and accelerometer data for each column in the input arrays.

    Parameters:
    AllfGyroX (ndarray): Array containing gyroscope X-axis data.
    AllfGyroY (ndarray): Array containing gyroscope Y-axis data.
    AllfGyroZ (ndarray): Array containing gyroscope Z-axis data.
    AllfAccelX (ndarray): Array containing accelerometer X-axis data.
    AllfAccelY (ndarray): Array containing accelerometer Y-axis data.
    AllfAccelZ (ndarray): Array containing accelerometer Z-axis data.
    """

    # Plot gyroscope data
    for i in range(AllfGyroX.shape[1]):
        plt.figure()
        plt.plot(AllfGyroX[:, i], 'r')  # Plot gyroscope X-axis data in red
        plt.plot(AllfGyroY[:, i], 'g')  # Plot gyroscope Y-axis data in green
        plt.plot(AllfGyroZ[:, i], 'b')  # Plot gyroscope Z-axis data in blue
        plt.xlabel('Time (s)')
        plt.ylabel('Gyroscope (°/s)')
        plt.xticks([126, 252, 378, 504], ['1', '2', '3', '4'])

        # Plot accelerometer data
        plt.figure()
        plt.plot(AllfAccelX[:, i], 'r')
        plt.plot(AllfAccelY[:, i], 'g')
        plt.plot(AllfAccelZ[:, i], 'b')
        plt.xlabel('Time (s)')
        plt.ylabel('Acceleration (g)')
        plt.xticks([126, 252, 378, 504], ['1', '2', '3', '4'])

    plt.show()
```
