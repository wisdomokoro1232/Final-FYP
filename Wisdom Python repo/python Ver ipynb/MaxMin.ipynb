```python
import numpy as np

def MaxMin(fGyroX, fGyroY, fGyroZ, fAccelX, fAccelY, fAccelZ):
    # Minimum
    gMin = [np.min(fGyroX), np.min(fGyroY), np.min(fGyroZ)]
    aMin = [np.min(fAccelX), np.min(fAccelY), np.min(fAccelZ)]

    # Maximum
    gMax = [np.max(fGyroX), np.max(fGyroY), np.max(fGyroZ)]
    aMax = [np.max(fAccelX), np.max(fAccelY), np.max(fAccelZ)]

    # Mean
    gMean = [np.mean(fGyroX), np.mean(fGyroY), np.mean(fGyroZ)]
    aMean = [np.mean(fAccelX), np.mean(fAccelY), np.mean(fAccelZ)]

    # Std
    gStd = [np.std(fGyroX), np.std(fGyroY), np.std(fGyroZ)]
    aStd = [np.std(fAccelX), np.std(fAccelY), np.std(fAccelZ)]

    # Under curve
    gArea = [np.trapz(fGyroX), np.trapz(fGyroY), np.trapz(fGyroZ)]
    aArea = [np.trapz(fAccelX), np.trapz(fAccelY), np.trapz(fAccelZ)]

    return aMax, aMin, aMean, aStd, aArea, gMax, gMin, gMean, gStd, gArea
```
